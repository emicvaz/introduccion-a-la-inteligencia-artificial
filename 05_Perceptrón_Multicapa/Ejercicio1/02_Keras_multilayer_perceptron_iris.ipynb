```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
target = iris.target
Y = to_categorical(target).astype("uint8")
# 0 => [1 0 0]
# 1 => [0 1 0]
# 2 => [0 0 1]
```

```python
idx = 0
print(X[idx])
print(target[idx])
print(Y[idx])
```

```python
model = keras.Sequential(
  [
    layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="sigmoid", name="layer2"),
  ]
)

model.summary()
```

```python
y = model(X)
print(y[149])
```

```python
optimizer=tf.keras.optimizers.SGD(learning_rate=0.03)
loss=tf.keras.losses.MeanSquaredError()

model.compile(optimizer, loss)
history = model.fit(X, Y, epochs=500)
```

```python
plt.plot(history.history['loss'])
plt.show()
```

```python
pd.DataFrame(history.history).plot(figsize=(8,5))
plt.show()
```

```python
model.predict(np.array([[3,3,1,1]]))
```

## Dos capas extra- Ejercicio 1

```python
# 1. Definición de la arquitectura profunda (4 capas Dense con activación sigmoide)
model_deep = keras.Sequential(
    [
        layers.Dense(3, activation="sigmoid", name="layer1", input_shape=(4,)),
        layers.Dense(3, activation="sigmoid", name="layer2"),
        layers.Dense(3, activation="sigmoid", name="layer3"),
        layers.Dense(3, activation="sigmoid", name="layer4"),
    ]
)

# 2. Resumen de la arquitectura (Evidencia obligatoria: debe mostrar 4 capas y 51 parámetros)
model_deep.summary()

```

```python
# 3. Verificación de salida inicial
y_deep = model_deep(X)
print("Salida inicial muestra 149 (red profunda):", y_deep[149].numpy())

```

```python
# 4. Compilación con idénticos hiperparámetros: SGD (lr=0.03) y MSE
optimizer_deep = tf.keras.optimizers.SGD(learning_rate=0.03)
loss_deep = tf.keras.losses.MeanSquaredError()

model_deep.compile(optimizer=optimizer_deep, loss=loss_deep)

# 5. Entrenamiento a 500 épocas
history_deep = model_deep.fit(X, Y, epochs=500, verbose=1)

```

```python
# 6. Gráfica comparativa directa de la pérdida (Loss / MSE)
plt.figure(figsize=(9, 5))
plt.plot(history.history['loss'], label='Keras Original (4x3x3)', color='blue', linewidth=2)
plt.plot(history_deep.history['loss'], label='Keras Profundo (4x3x3x3x3)', color='red', linestyle='--', linewidth=2)
plt.xlabel('Épocas')
plt.ylabel('Pérdida (MSE)')
plt.title('Comparación de Pérdida en Keras: Original vs Profundo')
plt.legend()
plt.grid(True)
plt.show()

```

```python
# 7. Comparación numérica de pérdida final
loss_orig_final = history.history['loss'][-1]
loss_deep_final = history_deep.history['loss'][-1]
print(f"MSE Final - Keras Original (4x3x3):         {loss_orig_final:.5f}")
print(f"MSE Final - Keras Profundo (4x3x3x3x3):     {loss_deep_final:.5f}")
```

```python

# 8. Predicción de ejemplo con la red profunda
sample_input = np.array([[3, 3, 1, 1]])
pred_orig = model.predict(sample_input)
pred_deep = model_deep.predict(sample_input)

print("\nPredicción para muestra [3, 3, 1, 1]:")
print("Modelo Original (4x3x3):     ", pred_orig)
print("Modelo Profundo (4x3x3x3x3): ", pred_deep)
```
